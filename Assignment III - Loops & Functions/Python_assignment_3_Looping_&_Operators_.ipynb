**Number Guessing Game**

```python
import random

# Generate random number between 1 and 10
secret_number = random.randint(1, 10)

name = input("Enter your name :")

print(f"Hello {name}.Welcome to Number Guessing Game!!")

#Ask the user to guess the number.
#number = int(input("Guess the number(Between 1 and 10):"))
#Set a variable attempts to 3,

attempts = 3

while attempts > 0 :
  number = int(input("Guess the number(Between 1 and 10):"))
  if number < 1 or number >10 :
      print("Your number is out of range .Please enter the number between 1 and 10 .")
      continue
  if number == secret_number:
      print ("Congradulations! You guessed the correct number.")
      break
  elif number > secret_number :
      print("Its too high .Try again .")
  else :
      print("Its too low .Try again .")
  attempts = attempts -1
else:
  print("Better luck next time.")

```

**For Loop - Multiplication Table Generator**

```python
#Prompt user for input.

print(f"**** Multiplication Table Generator ****")

number = int(input("Enter the number to generate Multiplication table : "))

#Generate multiplication table

for i in range(1,11):
  result = number * i
  print(f"{number} * {i} = {result}")
```

**Function- BMI Calculator**

```python
#Define a function calculate_bmi(weight, height)

print(f"**** BMI Calculator ****")

def calculate_bmi(weight, height):
    bmi = weight /(height ** 2)
    return bmi

weight = float(input("Enter your weight in kg :"))
height = float(input("Enter your height in metres :"))
bmi = calculate_bmi(weight, height)

print(f"Your BMI is :{bmi:.2f}")
```
