**List (Creation, Modification, and Access):**

1. List Creation:

```python
#a. Create a list named age_list with five integer elements.

age_list =[20,22,24,25,27]

#b. Create a list named name_list with five string elements.

name_list = ["Arun","Babu","Charu","Ravi","Siva"]

print(f"\nAge list: {age_list} \nName List : {name_list}")
```

**2.List operations / Modulations:**

```python
name_list = ["Arun","Babu","Charu","Ravi","Siva"]
age_list =[20,22,24,25,27]


#a. Append the string "Yazhini" to name_list.

name_list.append("Yazhini")
print("Append the string :",name_list)

#b. Insert the element 26 at index 2 in age_list.

age_list.insert(2,26)
print("Insert the element :",age_list)

#c. Remove the string "Yazhini" from name_list.

name_list.remove("Yazhini")
print("Removed the string :",name_list)

#d. Pop the last element from age_list.

age_list.pop()
print("Pop the last element :" ,age_list)

#e. Extend the age_list with additional ages [31,30,32].

age_list.extend([31,30,32])
print("Extended : ",age_list)

#f. Sort age_list in descending order.

age_list.sort(reverse=True)

print("Sort :",age_list)
#g. Find the max age, min age, and sum of all ages from age_list.

Max_age = max(age_list)
Min_age = min(age_list)
Sum_of_all_ages = sum(age_list)

print(f"\nMax age : {Max_age} \nMin age :{Min_age} \nSum of ages : {Sum_of_all_ages}")

```

**3. Accessing List Elements:**

```python
name_list = ["Arun","Babu","Charu","Ravi","Siva"]

print("\n ------ Listing Elements------")

#a. Print the first element of name_list.

print("First name :", name_list[0])

#b. Print the last element of name_list.

print("Last name :", name_list[-1])

#c. Print the elements from index 2 to index 4 in name_list.

print("Names from index 2 to 4 :", name_list[2:4])

#d. Print the elements of name_list in reverse order.

print("Reverse order of name list :", name_list[::-1])
```

**Dictionary (Creation, Modification and Access):**

```python

#a. Create a dictionary named student_marks that maps the names of five students to their marks (use a scale of 0 to 100).

student_marks = { "name":["Arun","Babu","Charu","Ravi","Siva"],
                 "marks":[85,90,95,75,98]}

print("Student details :",student_marks)

#b. Access and print the mark of a specific student of your choice.

print("\n***Second student details***")

print(f"\nName :",{student_marks["name"][2]},"\nMarks :",{student_marks["marks"][2]})


#c. Add a new student, "Janani," with a mark of 80 to the student_marks dictionary.

student_marks["name"].append("Janani")

student_marks["marks"].append(80)

print(f"\nNew Student :",student_marks)

#d. Update the mark of any one older student to 82.

student_marks["marks"][0] = 82

print("Updated marks of first student :",student_marks)

#e. Use the keys(), values(), and items() methods to print all keys, values, and key-value pairs in the student_marks dictionary.

print("\n****Students details****")

print(f"\nKeys :",student_marks.keys(),"\nValues :",student_marks.values(),"\nKey-Value pairs :",student_marks.items())
```

Performance Category Program:
1. Prompt the user for input. Score range should be from 0 to 10 (both inclusive).
2. Find the performance category based on the input score using the following criteria:
a. Above Average: Score greater than 7
b. Average: Score between 4 and 7 (both inclusive)
c. Below Average: Score less than 4
3. Output: Print the Performance category
4. Additional Step: You can give a prompt of your choice to each category. For example, if
the score is below average, “Need to improve your performance; consistent practice
will lead to better results”.

```python
score = int(input("Enter the score :"))

if score < 0 or score >=11:
  print("Invalid score.Enter the valid score between 0 and 10.")
elif score >7:
  print("Above Average","\nGood work and keep going")
elif score >= 4 and score <= 7:
    print("Average.","Need more practice")
else :
      print("Below Average","\nNeed to improve your performance; consistent practice will lead to better results")
```
